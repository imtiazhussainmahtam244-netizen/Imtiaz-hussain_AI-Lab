```python
!pip install -U google-genai

from google import genai

client = genai.Client(api_key="")


MODEL = "models/gemini-3.5-flash"


def self_driving_car_agent(user_input):

    prompt = f"""
You are an intelligent Self-Driving Car Agent.

Your job is to analyze the car's environment and make safe driving decisions.

Provide a well-structured response with clear headings:

1. Environment Analysis
   - Analyze road conditions
   - Detect obstacles
   - Check traffic lights
   - Consider pedestrians and other vehicles

2. Sensor Information
   - Distance sensor
   - Camera
   - GPS
   - Speed sensor
   - Traffic light detection

3. Driving Decision
   Choose the appropriate action:
   - Move Forward
   - Slow Down
   - Stop
   - Apply Brake
   - Turn Left
   - Turn Right

4. Reason for Decision
   Clearly explain why the car should take this action.

5. Safety Check
   - Maintain safe distance
   - Follow traffic rules
   - Avoid collisions
   - Give priority to pedestrians
   - Stop when danger is detected

Use simple and easy language suitable for students.

User Input:
{user_input}

Self-Driving Car Agent Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Self-Driving Car Intelligent Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = self_driving_car_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
